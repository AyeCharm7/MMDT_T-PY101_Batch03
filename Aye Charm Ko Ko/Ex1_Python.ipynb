# Ex 3: Temperature Converter

Write a program that convert temperature in Feranhite to celsius.

```python
# Get user input
celsius = float(input("Enter temperature in Celsius: "))

# Apply formula
fahrenheit = (celsius * 9/5) + 32

# Print
print("{}°C is equivalent to {}°F ".format(celsius, fahrenheit))
```

# Ex 4: Daily Maths

You are working in a retail shop. Write a program that computes the total cost for a customer that bought 3 differnt types of fruits : apple, orange, mango. Ask the user to enter the quantity and price for each type.

```python
# Apple
apple_quantity = float(input("Enter the quantity of apples: "))
apple_price = float(input("Enter the price per apple: "))

# Orange
orange_quantity = float(input("Enter the quantity of oranges: "))
orange_price = float(input("Enter the price per orange: "))

# Mango
mango_quantity = float(input("Enter the quantity of mangoes: "))
mango_price = float(input("Enter the price per mango: "))

# Calculate total cost
total_cost = (apple_quantity * apple_price) + \
             (orange_quantity * orange_price) + \
             (mango_quantity * mango_price)

print(f"\nThe total cost for the customer is: ${total_cost:.2f}")
```
