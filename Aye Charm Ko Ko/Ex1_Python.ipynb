# Ex 4: Daily Maths

You are working in a retail shop. Write a program that computes the total cost for a customer that bought 3 differnt types of fruits : apple, orange, mango. Ask the user to enter the quantity and price for each type.

```python
apple = float(input("Enter the quantity of apples: "))
price_apple = float(input("Enter the price per apple: "))

orange = float(input("Enter the quantity of oranges: "))
price_orange = float(input("Enter the price per orange: "))

mango = float(input("Enter the quantity of mangoes: "))
price_mango= float(input("Enter the price per mango: "))

total_cost = (apple * price_apple) + (orange * price_orange) + (mango * price_mango)

print(f"\nThe total cost for the customer is: ${total_cost:.2f}")
```

# Ex 6: Random Number
Write a program that generates a random number, x, between 1 and 50, a random number y between 2 and 5, and computes x^y.

```python
from random import randint
x = randint(1,50)
print('A random number between 1 and 50: ', x)
y = randint(2,5)
print('A random number between 2 and 5:', y)
result = x**y
print("x^y is: ", result)
```

# Ex 9: Computing

Write a program that asks the user for a number of seconds and prints out how many minutes and seconds that is. For instance, 200 seconds is 3 minutes and 20 seconds.

```python
total_seconds = int(input("Enter a number of seconds: "))

minutes = total_seconds // 60
remaining_seconds = total_seconds % 60

print("{} seconds is {} minutes and {} seconds.".format(total_seconds,minutes,remaining_seconds))
```

# Ex 10: Using Math Module

Write a program that asks the user for a number and then prints out the sine, cosine, and tangent of that number.

```python
import math

num = float(input("Enter a number:"))
sine_value = math.sin(num)
cosine_value = math.cos(num)
tangent_value = math.tan(num)

print("The sine of {} is: {}".format(num, sine_value))
print("The sine of {} is: {}".format(num, cosine_value))
print("The sine of {} is: {}".format(num, tangent_value))
```
