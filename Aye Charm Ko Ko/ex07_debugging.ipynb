```python
#Q4) The following program should check if a number is a prime number. Debug the code.

def is_prime(n):
    if n <= 1:
        return False
    for num in range(2, n):
        if n % num == 0:  # Corrected condition: check if n is divisible by num
            return False
    return True

numbers = np.arange(3, 21)
results = {}
for num in numbers:
    results[num] =is_prime(num)

print(f"Your results is {results}")
expected_result = {3: True, 4: False, 5: True, 6: False, 7: True, 8: False, 9: False, 10: False, 11: True, 12: False, 13: True, 14: False, 15: False, 16: False, 17: True, 18: False, 19: True, 20: False}
print(f"Expected result is: {expected_result}")
```

```python
#Q5) The following code uses stub functions to design the game. Complete the code.
import random

def generate_random_num():
    """
    Generates a random integer between 1 and 100 (inclusive) for the user to guess.
    """
    # random.randint(1, 100) returns a number N where 1 <= N <= 100
    return random.randint(1, 100)

def get_user_input():
    """
    This function asks the user to guess the number. Make sure that the number entered is an integer.
    """
    while True:
        try:
            guess_num = int(input("Guess a number between 1 and 100: "))
            return guess_num
        except ValueError:
            print("That's not a valid integer. Please try again.")

def check_the_number(actual_num, guess_num):
    """
    This function checks if the actual number and guess number are the same.
    """
    # Simply compare the two values and return the result of the comparison
    return actual_num == guess_num

def display_the_feedback(is_correct, chances):
    """
    This is a void function to print the feedback.
    """
    if is_correct:
        print("bingo")
    else:
        print(f"wrong guess. {chances} chances left.")

### Do not change the functions below ###
def guess_num_game():
    actual_num = generate_random_num()

    max_chances = 10
    for counter in range(1, max_chances+1):
        guess_num = get_user_input()
        is_correct = check_the_number(actual_num, guess_num)
        display_the_feedback(is_correct, max_chances - counter)
        if(is_correct):
            return "You did it! You won the game"

    return "You lost! Play one more time? "


def main():

    while True:
        win_lost_msg = guess_num_game()
        print(win_lost_msg)
        is_continue = int(input("Enter 1 to continue:"))
        if(not is_continue):
            return "Bye. Thanks for playing."

main()
```
